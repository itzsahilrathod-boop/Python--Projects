## 🚀 How to Run a Project
Because each project is stored in its own separate file/folder, you can run them individually:

1. Clone or download this repository to your local machine.
2. Navigate into the specific folder of the project you want to test.
3. Open the `.py` file in your preferred Python IDE (like PyCharm).
4. Run the file directly in your terminal:
   ```bash
   python main.py

## 🧠 Key Takeaways
This repository documents my early progress as I learn Python. Rather than relying on external libraries, these first 15 projects focus strictly on mastering **core Python**. Building these applications independently has strengthened my foundational skills in several key areas:

* **Mastering Control Flow:** Building text-based games like Hangman and Blackjack required deeply understanding `while` loops, nested `if/elif` statements, and managing continuous game states.
* **Modularizing Code with Functions:** As the projects grew in complexity (such as the Coffee Machine), I learned how to break down monolithic scripts into smaller, reusable functions with clear inputs and returns.
* **Data Structure Manipulation:** Projects like the Secret Auction and Caesar Cipher provided extensive practice in reading, writing, and updating Lists and Dictionaries.
* **Independent Debugging:** I improved my ability to trace logic errors and manage variable scope, learning to systematically hunt down bugs instead of just looking up the answers.
